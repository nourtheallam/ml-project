# Overview

**Input:** a dataset with an appended sentiment column.

**Output:** the same dataset with strings turned into numerical values.

```python
import pandas as pd
```

```python
# EDITABLE: import raw dataset, to be updated based on dataset's location
movies_dataset = pd.read_csv("./movies_dataset_updated.csv")
```

```python
# Split dataset into input features vs. values we are trying to predict
movies_dataset_x = (movies_dataset.drop(columns=["Rank", "Release Group", "$Worldwide", "$Domestic", "$Foreign", "Foreign %", "Domestic %"]))
movies_dataset_y = (movies_dataset[["$Domestic", "$Foreign"]])
# EDITABLE: drop the last column for now, until we have completed sentiment analysis
movies_dataset_x = movies_dataset_x.iloc[:, :-1]

# Convert all string values to numerical ones
# For Original_Language, this is easy because each row can only have one value
single_val_cols = ['Original_Language']
for col in single_val_cols:
    # Identify the different unique values this column/feature can have
    types = movies_dataset_x[col].unique()
    # Assign a number to each such value
    types_enum = {val: i for i, val in enumerate(types)}
    # Replace strings with numbers
    for t in types:
        movies_dataset_x[col] = movies_dataset_x[col].replace(t, types_enum[t])

# For Genres and Production Countries, we need to make a list of numbers instead
multi_val_cols = ['Genres', 'Production_Countries']
for col in multi_val_cols:
    # First, identify the types of entries we can have
    types = []
    for entry in movies_dataset_x[col].unique(): 
        if type(entry) == str:
            entry_types = entry.split(", ")
            for t in entry_types:
                if t not in types:
                    types.append(t)
    # Assign each type a number like we did before   
    types_enum = {val: i for i, val in enumerate(types)} 
    # Now, look at each individual entry and turn it into a list of numbers
    new_col = []
    for entry in movies_dataset_x[col]:
        entry_number_list = []
        if type(entry) == str:
            entry_types = entry.split(", ")
            for t in entry_types:
                entry_number_list.append(types_enum[t])
            new_col.append(entry_number_list)
        else:
            new_col.append([-1])
    movies_dataset_x[col] = new_col    

# For rating, just remove the "/10" term
movies_dataset_x["Rating"] = movies_dataset_x["Rating"].str[:-3]

# EDITABLE: add a step here for the sentiment analysis column

# Print the claned dataset
print(movies_dataset_x)

```
